# 02 - Candidate Generation & Blocking Evaluation

Benchmarks multi-index candidate blocking strategies and calculates the **blocking recall ceiling**.

```python
import sys
from pathlib import Path
sys.path.insert(0, "..")

from src.config import DATA_DIR_TRAIN
from src.load_data import load_dataset_split
from src.blocking import MultiIndexBlocker
from src.evaluate import evaluate_blocking
```

## 1. Load Data

```python
df_s1, df_s2, df_s3, gt = load_dataset_split(DATA_DIR_TRAIN, "train")
```

## 2. Run Multi-Index Blocking

```python
blocker = MultiIndexBlocker(tfidf_top_k=20, max_total_candidates=25)
candidates = blocker.generate_candidates(df_s1, df_s2, df_s3)
```

## 3. Evaluate Blocking Recall & Reduction Ratio

```python
cand_sets = {k: set(v) for k, v in candidates.items()}
metrics = evaluate_blocking(cand_sets, gt, len(df_s1), len(df_s2) + len(df_s3))

print(f"Blocking Recall Ceiling: {metrics['blocking_recall']*100:.2f}%")
print(f"Captured Links: {metrics['captured_links']} / {metrics['total_gt_links']}")
print(f"Reduction Ratio: {metrics['reduction_ratio']*100:.4f}%")
print(f"Average Candidates per S1: {metrics['avg_candidates_per_s1']:.2f}")
```
